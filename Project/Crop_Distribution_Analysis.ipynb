```python
import pandas as pd
import numpy as np

file_path = "/content/drive/MyDrive/Maharashtra_Agri_Master_2024_Jul2025.csv"

df = pd.read_csv(file_path)

print("Shape:", df.shape)
print("\nColumns:")
for i, col in enumerate(df.columns, 1):
    print(f"{i:02d}. {col}")
```

```python
commodity_dist = (
    df["Commodity"]
    .value_counts()
    .rename_axis("Commodity")
    .reset_index(name="Count")
)

commodity_dist["Percentage"] = (
    commodity_dist["Count"] / len(df) * 100
)

print(commodity_dist.to_string(index=False))
```

```python
print("Number of unique commodities:", df["Commodity"].nunique())
```

```python
crop_district = pd.crosstab(
    df["District"],
    df["Commodity"]
)

print("Shape:", crop_district.shape)
print(crop_district)
```

```python
crops_per_district = (
    df.groupby("District")["Commodity"]
    .nunique()
    .sort_values(ascending=False)
)

print(crops_per_district.to_string())
```

```python
district_crop_pct = pd.crosstab(
    df["District"],
    df["Commodity"],
    normalize="index"
) * 100

print(district_crop_pct.round(2))
```

```python
crops_per_market = (
    df.groupby("Market")["Commodity"]
    .nunique()
    .sort_values(ascending=False)
)

print(crops_per_market.to_string())
```
