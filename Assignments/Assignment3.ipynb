```python
import pandas as pd
df = pd.read_csv(r"C:\Users\91776\Downloads\online food delivery dataset.csv")


print("First 5 rows:")
print(df.head())

print("\nNumber of rows and columns:")
print(df.shape)


print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
average_age = df["Age"].mean()

minimum_age = df["Age"].min()

maximum_age = df["Age"].max()

male_customers = (df["Gender"] == "Male").sum()

female_customers = (df["Gender"] == "Female").sum()

print("Average Age:", average_age)
print("Minimum Age:", minimum_age)
print("Maximum Age:", maximum_age)
print("Number of Male Customers:", male_customers)
print("Number of Female Customers:", female_customers)
```

```python
filtered_customers = df[
    (df["Age"] > 25) &
    (df["Occupation"] == "Employee")
]
print(
    filtered_customers[
        ["Age", "Gender", "Occupation", "Monthly Income"]
    ]
)
```

```python
average_age = df.groupby("Occupation")["Age"].mean()

print("Average age for each occupation:")
print(average_age)

highest_occupation = average_age.idxmax()
highest_age = average_age.max()

print("\nOccupation with highest average customer age:")
print(highest_occupation)

print("Highest average age:", highest_age)
```

```python
feedback_count = df["Feedback"].value_counts()

print("Number of customers for each feedback:")
print(feedback_count)

positive_count = (df["Feedback"] == "Positive").sum()

total_customers = len(df)

positive_percentage = (positive_count / total_customers) * 100

print("\nTotal customers:", total_customers)
print("Positive feedback:", positive_count)
print("Percentage of positive feedback:", positive_percentage, "%")
```

```python
occupation_feedback = df.groupby(
    ["Occupation", "Feedback"]
).size()

print("Number of customers for each Occupation and Feedback:")
print(occupation_feedback)
```

```python
occupation_feedback = df.groupby(
    ["Occupation", "Feedback"]
).size().unstack(fill_value=0)

print("Occupation and Feedback Analysis:")
print(occupation_feedback)
```

```python
class BankAccount:

    def __init__(self, account, owner, balance):
        self.account = account
        self.owner = owner
        self.balance = balance
        
    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount

account1 = BankAccount("ACC001", "Sanika", 50000)

print("Initial Balance:", account1.balance)

account1.deposit(10000)
print("After Deposit:", account1.balance)

account1.withdraw(20000)
print("After Withdrawal:", account1.balance)
```

```python
class BankAccount:
    
    def __init__(self, account, owner, balance):
        self.account = account
        self.owner = owner
        self.balance = balance
        
    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):

        if amount <= self.balance:
            self.balance = self.balance - amount
            return True
        else:
            return False

account1 = BankAccount("ACC001", "Sanika", 50000)

result1 = account1.withdraw(10000)

print("Valid Withdrawal:")
print("Returned value:", result1)
print("Account Balance:", account1.balance)

result2 = account1.withdraw(50000)

print("\nInvalid Withdrawal:")
print("Returned value:", result2)
print("Account Balance:", account1.balance)
```

```python
class Student:

    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)
    
student1 = Student("Sanika", "MS2501", [80, 85, 90])
student2 = Student("Rahul", "MS2502", [70, 75, 80])
student3 = Student("Priya", "MS2503", [90, 85, 95])

print("Student:", student1.name)
print("Average Marks:", student1.average_marks())

print("\nStudent:", student2.name)
print("Average Marks:", student2.average_marks())

print("\nStudent:", student3.name)
print("Average Marks:", student3.average_marks())
```

```python
class Vehicle:
    def move(self):
        print("Vehicle is moving")

class Car(Vehicle):

    def move(self):
        print("Car is driving on the road")
class Boat(Vehicle):

    def move(self):
        print("Boat is sailing on the water")
vehicle = Vehicle()
car = Car()
boat = Boat()
vehicle.move()
car.move()
boat.move()
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, doors):
    
        super().__init__(brand)

        self.doors = doors

car1 = Car("Toyota", 4)
car2 = Car("Honda", 4)

print("Car 1")
print("Brand:", car1.brand)
print("Number of doors:", car1.doors)

print("\nCar 2")
print("Brand:", car2.brand)
print("Number of doors:", car2.doors)
```

```python
class Car:
    def move(self):
        print("Car is driving on the road")

class Boat:
    def move(self):
        print("Boat is sailing on the water")

class Airplane:
    def move(self):
        print("Airplane is flying in the sky")
car = Car()
boat = Boat()
airplane = Airplane()

vehicles = [car, boat, airplane]

for vehicle in vehicles:
    vehicle.move()
```
