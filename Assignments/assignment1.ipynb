```python
import numpy as np

def temperature_analysis(temp):

    average = np.mean(temp)
    maximum = np.max(temp)
    minimum = np.min(temp)

    highest_day = np.argmax(temp) + 1

    above_average = temp[temp > average]

    print("Average:", average)
    print("Maximum:", maximum)
    print("Minimum:", minimum)
    print("Highest Temperature Day:", highest_day)
    print("Above Average:", above_average)


temp = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])

temperature_analysis(temp)
```

```python
import numpy as np

def calculate(data):

    square = data ** 2
    root = np.sqrt(data)
    exponential = np.exp(data)

    print("Square:")
    print(square)

    print("Square root:")
    print(root)

    print("Exponential:")
    print(exponential)

    print("Mean of square:", np.mean(square))
    print("Mean of square root:", np.mean(root))
    print("Mean of exponential:", np.mean(exponential))


data = np.array([2, 4, 6, 8, 10, 12])

calculate(data)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([15, 25, 35, 45, 55])

# One array
one = np.concatenate((a, b))

# Separate rows
rows = np.vstack((a, b))

# Separate columns
columns = np.column_stack((a, b))

print("One-dimensional:")
print(one)
print("Shape:", one.shape)

print("\nRows:")
print(rows)
print("Shape:", rows.shape)

print("\nColumns:")
print(columns)
print("Shape:", columns.shape)
```

```python
import numpy as np

A = np.array([12, 15, 14, 18, 20, 17, 16])

B = np.array([10, 13, 15, 16, 19, 21, 18])

print("Mean A:", np.mean(A))
print("Mean B:", np.mean(B))

print("Standard deviation A:", np.std(A))
print("Standard deviation B:", np.std(B))

print("Maximum A:", np.max(A))
print("Maximum B:", np.max(B))

print("Minimum A:", np.min(A))
print("Minimum B:", np.min(B))

print("Difference:")
print(A - B)

if np.std(A) > np.std(B):
    print("A has greater variation")
else:
    print("B has greater variation")
```

```python
import numpy as np

def create_data():

    return np.random.randint(1, 101, (10, 5))


def analysis(a):

    print("Data:")
    print(a)

    print("\nShape:", a.shape)
    print("Dimensions:", a.ndim)

    print("\nMean of each variable:")
    print(np.mean(a, axis=0))

    print("\nMean of each experiment:")
    print(np.mean(a, axis=1))

    print("\nMaximum:", np.max(a))
    print("Minimum:", np.min(a))

    average = np.mean(a)

    print("\nOverall mean:", average)

    print("\nValues above mean:")
    print(a[a > average])

    b = a.reshape(5, 10)

    print("\nReshaped array:")
    print(b)

    print("\nFlattened:")
    print(b.flatten())


data = create_data()

analysis(data)
```

```python
import numpy as np

# Generate marks
marks = np.random.randint(0, 101, (10, 4))

print("Student Marks:")
print(marks)


# Average of each student
student_average = np.mean(marks, axis=1)

print("\nStudent Averages:")
print(student_average)


# Average of each subject
subject_average = np.mean(marks, axis=0)

print("\nSubject Averages:")
print(subject_average)


# Highest average student
best_student = np.argmax(student_average) + 1

print("\nStudent with highest average:",
      best_student)


# Lowest average student
worst_student = np.argmin(student_average) + 1

print("Student with lowest average:",
      worst_student)


# Highest mark
print("Highest mark:",
      np.max(marks))


# Lowest mark
print("Lowest mark:",
      np.min(marks))


# Students with average above 75
count = np.sum(student_average > 75)

print("Students with average above 75:",
      count)


# Marks greater than 90
print("Marks greater than 90:")
print(marks[marks > 90])


# Grades
def grade(avg):

    if avg >= 90:
        return "A+"
    elif avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"


print("\nStudent Grades:")

for i in range(10):

    print(
        "Student", i + 1,
        "Average =", round(student_average[i], 2),
        "Grade =", grade(student_average[i])
    )
```
