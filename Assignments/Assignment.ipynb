```python
# Experiment 2: String Indexing

text = "Python"

print("Original String:", text)

print("First Character :", text[0])
print("Fourth Character:", text[3])
print("Last Character  :", text[-1])
```

```python
# Experiment 3: String Slicing

text = "Programming"

print("Original String:", text)

print("text[0:5] =", text[0:5])
print("text[3:]  =", text[3:])
print("text[:6]  =", text[:6])
print("text[::-1]=", text[::-1])
```

```python
# Experiment 4: List Operations

numbers = [10, 20, 30]

print("Original List:", numbers)

numbers.append(40)
print("After append:", numbers)

numbers.remove(20)
print("After remove:", numbers)
```

```python
# Experiment 5: Type Conversion

age = input("Enter your age: ")

print("Without Conversion:", age + age)

age = int(age)

print("After Conversion:", age + age)
```

```python
# Experiment 6: List Indexing

fruits = ["Apple", "Banana", "Mango", "Orange"]

print("List:", fruits)

print("First Fruit :", fruits[0])
print("Last Fruit  :", fruits[-1])
print("Slice       :", fruits[1:3])
```

```python
# Experiment 7: Mutable List

numbers = [10, 20, 30]

print("Original List:", numbers)

numbers[1] = 200

print("Updated List :", numbers)
```
